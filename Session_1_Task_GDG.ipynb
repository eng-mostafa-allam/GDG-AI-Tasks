*Learn how to use:*

---

#### **Task:**
> Ask the user to enter a number (integer).
> - If the number is greater than 0 AND is even -> Print "Positive and Even"
> - If the number is greater than 0 AND is odd -> Print "Positive and Odd"

## <font color="#4285F4">Task 5: Registration System (Comprehensive)</font>
#### **Objective:**
*Combine Variables, Strings, Data Structures, and Conditionals.*

---

#### **Task:**
> Create a smart registration system:
> 1. Ask the user for their **Name**, **Age** (integer), and **Email**.
> 2. **Clean the data**: Make the name `.title()` and `.strip()` extra spaces. Make the email `.lower()`.
> 3. **Check Condition**:
>    - If the user is 18 or older:
>      - Ask them to enter 3 skills separated by spaces (e.g., "Python SQL Math").
>      - `.split()` the skills into a List.
>      - Create a Dictionary saving their cleaned Name, Age, Email, and Skills list.
>      - Print: "Registration Successful!" and print the dictionary.
>    - If they are under 18:
>      - Print: "Sorry, you must be 18 or older to register."

---
#### **Example Output (Age >= 18):**
```text
Enter Name:    aHmEd ali   
Enter Age: 20
Enter Email: AHMED@gmail.com
Enter 3 skills separated by space: AI Python SQL

Registration Successful!
User Data: {'Name': 'Ahmed Ali', 'Age': 20, 'Email': 'ahmed@gmail.com', 'Skills': ['AI', 'Python', 'SQL']}
```

````python
name = input("Enter Name: ")
age = int(input("Enter Age: "))
email = input("Enter Email: ")

name = name.strip().title()
email = email.lower()

if age >= 18:
    skills = input("Enter 3 skills separated by space: ")
    skills_list = skills.split()

    user_data = {'Name': name, 'Age': age, 'Email': email,'Skills': skills_list}

    print("\nRegistration Successful!")
    print("User Data:",user_data)
else:
    print("Sorry, you must be 18 or older to register.")
````

## <font color="#E37400">Bonus Task 1: The Time Alchemist</font>
#### **Objective:**
*Advanced use of:*
- Arithmetic Operators (`//` Floor Division, `%` Modulus)
- Conditional Formatting

---

#### **Task:**
> Write a program that asks the user to enter a huge amount of **Seconds** (integer).
> Your job is to convert this number into a readable format: **Days, Hours, Minutes, and remaining Seconds**.
>
> *Hint: 1 Minute = 60 Seconds, 1 Hour = 3600 Seconds, 1 Day = 86400 Seconds.*
>
> **Extra Challenge**: Only print the non-zero values! (If days = 0, don't print days).

---
#### **Example Output:**
```text
Enter time in seconds: 100000

--- Time Conversion ---
1 Days
3 Hours
46 Minutes
40 Seconds
```

````python
total_seconds = int(input("Enter time in seconds: "))

days = total_seconds // 86400
remain = total_seconds % 86400

hours = remain // 3600
remain = remain % 3600

minutes = remain // 60
seconds = remain % 60

print("\n" + " Time Conversion ".center(23, '-'))

if days > 0:
    print(days,"Days")
if hours > 0:
    print(hours,"Hours")
if minutes > 0:
    print(minutes,"Minutes")
if seconds > 0:
    print(seconds,"Seconds")
````

## <font color="#34A853">Bonus Task 2: The Maximum Subarray</font>
#### **Objective:**
*Algorithmic Problem Solving & Critical Thinking*

---

#### **Task:**
> **Context:** You are given a list of integers (both positive and negative) representing the daily profit and loss of a company.
>
> `profits = [-2, 1, -3, 4, -1, 2, 1, -5, 4]`
>
> **Your Mission:** Find the **maximum sum** of any contiguous sub-sequence (a continuous part of the list).
>
> *How will you solve it? The stage is yours. Think like a real software engineer and show us your logic!*

---
#### **Example Output:**
```text
Daily Profits: [-2, 1, -3, 4, -1, 2, 1, -5, 4]

The Maximum Contiguous Profit is: 6
(Explanation: The sub-sequence [4, -1, 2, 1] gives the maximum sum of 6)
```

````python
DailyProfits = [-2, 1, -3, 4, -1, 2, 1, -5, 4]
print("Daily Profits:",DailyProfits) #to be exact like Example Output(^_^)

n = len(DailyProfits)
lst = list()

for i in range(n):
  count = 0

  for j in range(i , n):
    count += DailyProfits[j]
    lst.append(count)
print("The Maximum Contiguous Profit is:" , max(lst))
````

```text

Is 11 Prime? True
Prime numbers before 11: [2, 3, 5, 7]
```

````python
num = int(input("Enter a number: "))

li = list()

if num < 2 :
  check1 = False
else:
  check1 = True
  for i in range(2 , num):

    if num%i == 0 :
      check1 = False

    check2 = True
    for j in range(2,i):
      if i%j == 0 :
        check2 = False
    if check2:
      li.append(i)

print("\nIs", num ,"Prime?",check1)
print(f"Prime numbers before {num}: ",li)
````
