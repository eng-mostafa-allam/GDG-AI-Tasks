## **Task 8: Detect Outliers (IQR)**

#### **Objective:**
*Learn how to use:*
- `quantile()`
- The IQR rule
- Writing your own function with `def`

 *Need help? Go back to the presentation: **Outliers (both slides)***

---

#### **Task:**
> 1. Write a function called `iqr_outliers` that takes a DataFrame and a column name, and returns the outlier rows, the lower bound and the upper bound.
> 2. Use it on `Age` and on `Fare`, and print the bounds and the number of outliers for each.
> 3. *(Optional)* Draw a boxplot of `Fare`.
> 4. **Think:** Must every outlier be removed? Why might a very high `Fare` be perfectly normal?

---
#### **Example Output:**
```text
Age  -> bounds: (2.5, 54.5) | outliers: 66
Fare -> bounds: (-26.72, 65.63) | outliers: 116
```


````python
def iqr_outliers(df, column):
  Q1 = df[column].quantile(0.25)
  Q3 = df[column].quantile(0.75)
  IQR = Q3 - Q1

  lower = Q1 - 1.5 * IQR
  upper= Q3 + 1.5 * IQR

  outliers = df[(df[column] < lower) | (df[column] > upper)]
  print(f"{column}  -> bounds: ({round(lower, 2)}, {round(upper, 2)}) | outliers: {len(outliers)}")

iqr_outliers(df_clean , "Age")
iqr_outliers(df_clean , "Fare")
````

## **Task 9: Handle Outliers**

#### **Objective:**
*Learn how to use:*
- Filtering rows with conditions
- `clip()`

 *Need help? Go back to the presentation: **Outliers***

---

#### **Task:**
> 1. In `Age`: **remove** the outlier rows.
> 2. In `Fare`: **don't remove** them. Cap the extreme values at the bounds instead (recalculate the bounds after removing the `Age` outliers).
> 3. Print the new shape and the min/max of `Fare`.
> 4. **Think:** What's the difference between removing and clipping? When would you use each?

---
#### **Example Output:**
```text
Shape after removing Age outliers: (825, 11)
Fare after clipping -> min: 0.0, max: 63.33
```


````python
Q1_age = df_clean["Age"].quantile(0.25)
Q3_age = df_clean["Age"].quantile(0.75)
IQR_age = Q3_age - Q1_age

lower_age = Q1_age - 1.5 * IQR_age
upper_age = Q3_age + 1.5 * IQR_age


df_clean = df_clean[((df_clean["Age"] >= lower_age) & (df_clean["Age"] <= upper_age)) | df_clean["Age"].isna()]

Q1_fare = df_clean["Fare"].quantile(0.25)
Q3_fare = df_clean["Fare"].quantile(0.75)
IQR_fare = Q3_fare - Q1_fare

lower_fare = Q1_fare - 1.5 * IQR_fare
upper_fare = Q3_fare + 1.5 * IQR_fare

df_clean["Fare"] = df_clean["Fare"].clip(lower=lower_fare, upper=upper_fare)

print(f"Shape after removing Age outliers: {df_clean.shape}")
print(f"Fare after clipping -> min: {df_clean['Fare'].min()}, max: {round(df_clean['Fare'].max(), 2)}")
````
